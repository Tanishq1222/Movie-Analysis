### EDA (Exploratory Data Analysis)

```python
import pandas as pd, numpy as np 
```

```python
pd.set_option('display.max_columns',None)
```

```python
df = pd.read_csv('cleaned_movie_dataset.csv')
```

```python
df.head()
```

```python
# Whcih Genre is most profitable on an average?

genre_avg = df.groupby('Genre',observed = True ).agg(
    avg_profit=('ProfitUSD','mean'),
    avg_ROI = ('ROI_pct','mean'),
    count = ('MovieID','count')    
).round(2).sort_values('avg_ROI',  ascending = False)
print("_____Most Profitable Genre (by Average ROI%)______ ")
print(genre_avg)
```

**Genre profitability** — Horror leads with the highest ROI% (178.64%), followed by Action (178.40%) and Comedy (177.93%). However, the spread is very narrow, with all genres hovering between 177% and 179% ROI.

```python
# Does budget correlate with box office?
corr_budget_USbo = df['BudgetUSD'].corr(df['US_BoxOfficeUSD'])
corr_budget_Gbo =  df['BudgetUSD'].corr(df['Global_BoxOfficeUSD'])
corr_budget_RT = df['BudgetUSD'].corr(df['RottenTomatoesScore'])
corr_budget_IMDb =  df['BudgetUSD'].corr(df['IMDbRating'])
print("---Budget Correlation---")
print(f"  Budget -> US BO:       {corr_budget_USbo:.4f}")
print(f"  Budget -> Global BO:   {corr_budget_Gbo:.4f}")
print(f"  Budget -> RT Rating:   {corr_budget_RT:.4f}")
print(f"  Budget -> IMDb Rating: {corr_budget_IMDb:.4f}")
```

**Budget Correlation** — Budget and Global Box Office have very strong correlation of 0.8955 - Higher budget almost means higher box office revenue 

```python
# Which Decade had best avg ROI?

decade_avg = df.groupby('Decade', observed = True).agg(
    avg_roi = ('ROI_pct','mean')
).round(2).sort_values('avg_roi',ascending = False)
print("---Best ROI by Decade---")
print(decade_avg)
```

**Best Decade** — The 1950s had the highest Avg ROI (178.47%). But, the difference across all decade is marginal. No single era dramatically outperform another

```python
# Does opening day predict total box office?

corr_opening_USbo = df['Opening_Day_SalesUSD'].corr(df['US_BoxOfficeUSD'])
corr_opening_Gbo = df['Opening_Day_SalesUSD'].corr(df['Global_BoxOfficeUSD'])
print('---Opening day correlation---')
print(f'Opening day -> US BO :     {corr_opening_USbo:.3f}')
print(f'Opening Day -> Global BO : {corr_opening_Gbo:.3f}')
```

**Opening Day Correlation** — Opening day Sales vs Global Box Office correlation is *0.941* - is the strongest relationship in the entire dataset. A movies's very first day of sales is the best predictor of its total lifetime earning.

```python
# Which country produces high rated movies?

HR_movies = df.groupby('Country' , observed = True).agg(
    avg_imdb = ('IMDbRating','mean'),
    count=('MovieID','count')
).round(2).sort_values('avg_imdb',ascending= False)

print("---IMDb rating by country")
print(HR_movies)
```

**Country And Rating** — Japan,Australia,UK produces highest-rated movies on average (6.51,6.50,6.50 IMDb). But the gap between best and worst is 0.03 only.

```python
# Top 10 Director by Average ROI

d_avg = df.groupby('Director').agg(
    Avg_ROI = ('ROI_pct','mean'),
    Count = ('MovieID','count')
).round(2).sort_values('Avg_ROI',ascending = False).head(10)
print('---Top 10 Director by Average ROI---')
print(d_avg)
```

**Top Director By ROI** — Kim Watkins, Karen Smith, Chales Lopez lead on avg ROI with 182.1%, 181.8%, 181.4% respectively - each with 6500+ movies.

```python
# IMDb VS RT Correlation

corr_imdb_rt = df['IMDbRating'].corr(df['RottenTomatoesScore'])
print(f'IMDB Rating -> RT Correlation score : {corr_imdb_rt:.3f}')
```

**IMDb Vs RT Correlation** — Despite having different voting mechanisms (public votes vs. critic reviews), the two platforms share a strong correlation (0.829). If a movie is good, both tend to agree with it.

```python
# Box Office Trend over Decade

B_D = df.groupby('Decade', observed=True)['Global_BoxOfficeUSD'].median().round(0)
print(f"---Median global box office by decade---")
print(B_D)
```

**Box office over decades** — Median global box office is remarkably stable across all decades (~$8M)

#### Biggest Insight
* The single strongest business signal in this dataset is - **Opening Day Sales predict Total Box Ofiice with 0.941 correlation**. If a studio want to forecast a movie's lifetime revenue, they need to wait for one day.
